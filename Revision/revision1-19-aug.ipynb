#### VALID PLANDROM


```python
word = "madam"

def is_valid_plandrom(word):
    i=0
    j=len(word)-1

    while i<j:
        if word[i] == word[j]:
            i +=1
            j-= 1
        else:
            return False
        
    return True

is_valid_plandrom(word)
```

#### Valid Anagram

```python
word1 = "madam"
word2 = "dadam"

def is_valid_anagram(word1, word2):
    if len(word1) != len(word2):
        return False
    word1 = sorted(word1)
    word2 = sorted(word2)
    if word1 == word2:
        return True
    return False
    

is_valid_anagram(word1,word2)
```

#### longest common prefiix

```python
word1 = "fliwer"
word2 = "flight"

def longest_prefix(word1, word2):
    result = []
    w1_len = len(word1)
    w2_len = len(word2) 
    for i in range(min(w1_len,w2_len)):
        if word1[i] == word2[i]:
            result.append(word1[i])
        else:
            break
    return result
    

print(longest_prefix(word1,word2))
```

## rotate array 

```python
nums =[1,2,3,4,5,6,7]
k = 3

def rotate(nums, start, end ):
    i = start
    j =  end 

    while(i<j):
        nums[i],nums[j] = nums[j],nums[i]
        i += 1
        j -= 1

    return nums



print(rotate(nums, 0 , len(nums)-1))
print(rotate(nums, 0, k-1) )
print(rotate(nums, k , len(nums)-1))
```

## move zero 

```python
nums = [0,1,0,3,12]

k = 0 
for num in nums:
    if num == 0:
        continue
    else:
        nums[k] =  num 
        k += 1
print(nums)

while(k < len(nums)):
    nums[k] = 0 
    k += 1 
print(nums)
```
